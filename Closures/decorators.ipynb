```python
def counter(fn):
    count = 1
    def inner(*args, **kwargs):
        nonlocal count 
        count +=1 
        print(f"Function {fn.__name__} was called {count} times!")
        return fn(*args, **kwargs)
    return inner 
        
```

```python
def add(a, b):
    return a+b

```

```python
#conventional way to decorate a function
add = counter(add)
add(1,2)
```

```python
#modern way to decorate a function
@counter
def mul(a,b):
    return a*b

#this is same as what we did above
# mul = counter(mul)
```

```python
mul(10, 20)
```

```python
#timer decorator
def timed(fn):
    from time import perf_counter
    from functools import wraps
    @wraps(fn)                              
    def inner(*args, **kwargs):
        start = perf_counter()
        result = fn(*args, **kwargs)
        end = perf_counter()
        print(f"Function {fn.__name__} took {end - start:.6f} seconds to execute.")
        return result
    return inner
        
```

```python
@timed
def fib(n):
    if n <= 2:
        return 1
    fib_1, fib_2 = 1,1
    for _ in range(3, n+1):
        fib_2, fib_1 = fib_1 + fib_2, fib_2
    return fib_2
        

    
```

```python
fib(3)
```

```python
fib(1000)
```

```python
# log decorator
def log(fn):
    from functools import wraps
    from datetime import datetime, timezone
    @wraps(fn)
    def inner(*args, **kwargs):
        print(f"Calling function {fn.__name__} with arguments {args} and keyword arguments {kwargs} at {datetime.now(timezone.utc).isoformat()}")
        result = fn(*args, **kwargs)
        return result
    return inner
```

```python
@timed
@log
def compute_square(n):
    return n * n
```

```python
compute_square(10)
```

MEMOIZATION WITH DECORATORS

```python
def memoize(fn):
    cache = dict()
    
    def inner(n):
        if n in cache:
            return cache[n]
        res = fn(n)
        cache[n] = res
        return res
    return inner
```

```python
def fib(n):
    print("Computing fib(", n, ")")
    return 1 if n <= 2 else fib(n-1) + fib(n-2)
```

Basic Closure
Write a function make_multiplier(n) that returns a closure which multiplies its argument by n.

```python
def make_multiplier(n):
    
    def inner(number):
        return number*n
    return inner
```

```python
times5 = make_multiplier(5)
times7 = make_multiplier(7)
print(times5(10))
print(times5(12))
print(times7(12))
print(times7(8))
```

 Simple Decorator
Create a decorator @uppercase that converts the string returned by a function to uppercase.

```python
def uppercase(fn):
    def inner(*args, **kwargs):
        res = fn(*args, **kwargs)
        res = res.upper()
        return res 
    return inner 

@uppercase
def greet(name):
    return f"Hello {name}"
```

```python
greet("prasad")
```

Counting Calls
Write a decorator @count_calls that counts how many times a function has been called and prints the count each time.

```python
import random

def count_calls(fn):
    counts = 0
    def inner(*args, **kwargs):
        nonlocal counts
        counts +=1 
        res = fn(*args, **kwargs)
        print(f"{fn.__name__} was called {counts} times!")
        return res
    return inner 

@count_calls
def greet(name):
    return f"Hello, {name}!"

for _ in range(10):
    print(greet(random.choice(["Prasad", "Sunil", "Suneet", "Sushant", "Ankita"])))
```

Closure with Mutable State
Implement a function make_accumulator() that returns a closure maintaining a running sum. Each call to the closure should add the argument to the sum and return the new total.

```python
def make_accumulator():
    running_sum = 0
    
    def inner(num):
        nonlocal running_sum
        running_sum += num 
        return running_sum
    return inner 
```

```python
fn_1 = make_accumulator()
```

Create a decorator @repeat(n) that makes a function execute n times.

```python
def repeat(n):
    
    def dec(fn):
        
        def inner(*args, **kwargs):
            for _ in range(n):
                fn(*args, **kwargs)
                print(f"Executed {fn.__name__} {_} time!")
        return inner 
    return dec 
```

```python
@repeat(10)
def greet():
    print("Hello World!")
```

```python
greet()
```

Write a decorator @timer that measures and prints how long a function takes to execute, preserving the function's return value.

```python
def timer(fn):
    from time import perf_counter
    
    def inner(*args, **kwargs):
        start = perf_counter()
        res = fn(*args, **kwargs)
        end = perf_counter()
        print(f"{fn.__name__} took {end - start} time to run!")
        return res 
    return inner 
```

```python
@timer 
def greet(name):
    return f"Hello, {name}!"
```

```python
print(greet("Prasad"))
print(greet("Akansha"))
print(greet("Sovan"))
print(greet("Soumya"))
```

Implement a @memoize decorator that caches function results based on arguments to avoid redundant calculations (works with hashable arguments only).

```python
def memoize(fn):
    cache = dict()
    
    def inner(n):
        if n in cache:
            return cache[n]
        print(f"Calculating {fn.__name__} for {n}")
        res = fn(n)
        cache[n] = res
        return res
    return inner
```

```python
@memoize
def fib(n):
    if n==1 or n==2:
        return 1
    return fib(n-1) + fib(n-2) 
    
```

```python
fib(10)
```

Create a decorator factory @validate_range(min_val, max_val) that checks if a function's return value falls within the specified range, raising a ValueError if not.

```python
def validate_range(min_val, max_val):
    
    def dec(fn):
        
        def inner(*args, **kwargs):
            res = fn(*args, **kwargs)
            if  not min_val<=res<=max_val:
                raise ValueError("Output beyond the correct range!")
            return res 
        return inner 
    return dec 

                
```

```python
@validate_range(1, 100)
def mul(a,b):
    return a*b

''' 
what you are essentailly doing here is 
mul = validate_range(1,100)(mul)
so now, mul is the inner function of the decorator which when you call, it runs the inner
'''
    
```

```python
mul(100,0.5)
```

```python
mul = validate_range(1, 100)(mul)
```
