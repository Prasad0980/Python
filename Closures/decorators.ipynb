```python
def counter(fn):
    count = 1
    def inner(*args, **kwargs):
        nonlocal count 
        count +=1 
        print(f"Function {fn.__name__} was called {count} times!")
        return fn(*args, **kwargs)
    return inner 
        
```

```python
def add(a, b):
    return a+b

```

```python
#conventional way to decorate a function
add = counter(add)
add(1,2)
```

```python
#modern way to decorate a function
@counter
def mul(a,b):
    return a*b

#this is same as what we did above
# mul = counter(mul)
```

```python
mul(10, 20)
```
