1. Basic Counter
Create a function make_counter() that returns a counter function. Each time you call the returned function, it should return the next number in sequence starting from 1.

```python
def make_counter():
    count = 0
    
    def counter():
        nonlocal count 
        count +=1 
        return count 

    return counter
```

```python
counter = make_counter()
```

```python
print(counter())
print(counter())
print(counter())
```

2. Customizable Multiplier
Write a function make_multiplier(n) that returns a function which multiplies its argument by n. The value of n should be "remembered" by the returned function.

```python
def make_multiplier(n):
    
    def number(num):
        nonlocal n
        return num*n 
    return number 
```

```python
times2 = make_multiplier(2)
times3 = make_multiplier(3)
print(times2(5))
print(times2(56))
print(times3(10))
print(times3(18))
```

3: Running Average Calculator
Create a function make_averager() that returns a function to calculate the running average of all numbers passed to it. The returned function should maintain a history of all numbers it has seen.

```python
def make_averager():
    total, count = 0, 0
    def avg(num):
        nonlocal total, count 
        total += num 
        count +=1 
        return total/count
    return avg 
```

```python
num_avg = make_averager()
```

```python
num_avg(1)
```

```python
num_avg(9)
```

```python
num_avg(8)
```

Rate Limiter
Write a function make_rate_limiter(max_calls, time_window) that returns a decorated version of any function that limits how many times it can be called within a time window (in seconds). If the limit is exceeded, it should return None instead of executing the function.

```python
from time import perf_counter

def make_rate_limiter(max_calls, time_window):
    def decorator(fn):
        call_count = 0
        window_start = None
        
        def inner(*args, **kwargs):
            nonlocal call_count, window_start
            current_time = perf_counter()
            
            if window_start is None or current_time - window_start > time_window:
                window_start = current_time
                call_count = 0

            if call_count < max_calls:
                call_count += 1
                return fn(*args, **kwargs)
            else:
                return None
        
        return inner
    return decorator

@make_rate_limiter(5, 10)
def greet(name):
    return f"Hello {name}!"
```

```python
print(greet("Prasad"))
print(greet("Prasad"))
print(greet("Prasad"))
print(greet("Prasad"))
print(greet("Prasad"))
print(greet("Prasad"))
print(greet("Prasad"))
```
